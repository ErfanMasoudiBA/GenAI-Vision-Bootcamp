# Curse of Dimensionality

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

dimensions = list(range(2, 51))
distances = []

for d in dimensions:
    
    x = np.random.randint(0, 100, size=d)
    y = np.random.randint(0, 100, size=d)

    dist = np.linalg.norm(x - y)  # همان sqrt(sum((x_i - y_i)^2))
    distances.append(dist)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(dimensions, distances, marker='o', color='purple', linewidth=2, markersize=5)
plt.title("Euclidean Distance vs. Number of Dimensions (Curse of Dimensionality)")
plt.xlabel("Dimension (d)")
plt.ylabel("Euclidean Distance")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
