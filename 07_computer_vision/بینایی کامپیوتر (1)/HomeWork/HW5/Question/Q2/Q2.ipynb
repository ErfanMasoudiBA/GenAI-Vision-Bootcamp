##Read Images

```python
import cv2
from matplotlib import pyplot as plt

path=['messi_1.jpg', 'messi_2.jpg', 'messi_3.jpg','ronaldo_1.jpg', 'ronaldo_2.jpg', 'ronaldo_3.jpg']

allimage =[]
for p in path:
  img = cv2.imread(p)
    
  img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
  face=img_rgb.copy()
  face_data = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_alt.xml')
  found = face_data.detectMultiScale(img_gray, minSize =(20, 20))
  
  amount_found = len(found)
  if amount_found != 0:
      for (x, y, width, height) in found:
          cv2.rectangle(img_rgb, (x, y), 
                        (x + height, y + width), 
                        (0, 255, 0), 5)
          
          face = face[y:y+height,x:x+width]
          allimage.append(face)
          
  plt.subplot(1, 1, 1)
  plt.imshow(face)
  plt.show()
  plt.figure()
  
```
