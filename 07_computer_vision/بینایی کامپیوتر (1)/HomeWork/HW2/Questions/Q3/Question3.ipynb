### گام دوم
نقاط مربوط به هر تصویر را روی آن نمایش دهید و با مشاهده نقاط، 8 مورد از بهترین نقاط متناظر را به دلخواه انتخاب کنید. از تابع آماده زیر برای نمایش نقاط روی تصاویر استفاده کنید. در ورودی به ترتیب دو تصویر ورودی و سپس نقاط متناظر را با ترتیب وارد کنید

```python
def drawlines(img1,img2,lines,pts1,pts2, select):
    r,c = img1.shape[:2]
    img1 = cv2.cvtColor(img1,cv2.COLOR_GRAY2BGR)
    img2 = cv2.cvtColor(img2,cv2.COLOR_GRAY2BGR)
    for r,pt1,pt2 in zip(lines,pts1,pts2):
        color = tuple(np.random.randint(0,255,3).tolist())
        x0,y0 = map(int, [0, -r[2]/r[1] ])
        x1,y1 = map(int, [c, -(r[2]+r[0]*c)/r[1] ])
        if select :
          img1 = cv2.line(img1, (x0,y0), (x1,y1), color, 8)
        img1 = cv2.circle(img1,tuple(pt1),5,color, thickness = 30)
        img2 = cv2.circle(img2,tuple(pt2),5,color, thickness = 30)
    return img1,img2
```

```python
# با دستور زیر تصاویر را نمایش دهید
out1, out2 = drawlines(image1,image2,points1,points2, 0)

# هشت مورد از نقاز متناظر را به دلخواه وارد کنید
pts2 = np.array([[0, 0], [0, 0], [0, 0],[0, 0],   # نقاط انتخابی تصویر اول. به جای صفر های وارد شده مختصاب مناسب وارد کنید
              [0, 0], [0, 0], [0, 0],[0, 0])     # دقت کنید که بعد سوم مختصات را همواره عدد یک بماند چرا که در فضای همگن هستیم


pts1 = np.array([[0,  0], [0, 0], [0, 0],[0, 0],   # نقاط انتخابی تصویر دوم. به جای صفر های وارد شده مختصاب مناسب وارد کنید
              [0, 0], [0, 0], [0, 0],[0, 0]])      # دقت کنید که بعد سوم مختصات را همواره عدد یک بماند چرا که در فضای همگن هستیم
```
