````python
model = YOLO('yolov8n.pt')
image_path = 'Equine-Blog-A-dog-spooked-my-horse-Feature-image-900x507.png'
results = model.predict(image_path, conf=0.25)
````

````python
import cv2 
import matplotlib.pyplot as plt

print(results[0].boxes.xyxy)
print(results[0].boxes.conf)
print(results[0].boxes.cls)

img = cv2.imread(image_path)
for r in results:
    for i, box in enumerate(r.boxes.xyxy):
        x1, y1, x2, y2 = map(int, box) # Convert coordinates to integers
        label = f'{r.names[int(r.boxes.cls[i])]} {r.boxes.conf[i]:.2f}'

        # Draw bounding box
        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2) # Green box, thickness 2

        # Add label and confidence score
        cv2.putText(img, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

# Convert BGR image to RGB for Matplotlib display
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display the image with Matplotlib
plt.imshow(img_rgb)
plt.axis('off') # Hide axes
plt.show()
````

<div dir="rtl" style="text-align: right;">
<h3 style="text-align: right;">
تمرین:
</h3>

- برای اینکه آموزش سریعتر شود داده ولیدیشن را برای آموزش و داده تست را برای اعتبارسنجی استفاده کنید
-  هایپرپارامترهای مدل آلترالتیکس مانند نرخ یادگیری، مومنتم، نوع بهینه‌ساز را تغییر دهید و بهترین نتیجه را انتخاب کنید
</div>

````python
from ultralytics import YOLO
import random
import math
import time
data_location = "../notebooks-data/tyre-2"
for k in range(10):
    random.seed(time.time() + k * 1000)  # Different seed each time
    params = {
    'augment': random.choice([True, False]),
    'optimizer': random.choice(['SGD', 'Adam', 'AdamW', 'RMSprop']),
    'lr0': math.pow(10, random.uniform(-4, -2)),
    'momentum': random.uniform(0.85, 0.98),
    'weight_decay': math.pow(10, random.uniform(-5, -3)),
    'warmup_epochs': random.choice([2, 3, 4, 5]),
    'close_mosaic': random.choice([0, 1, 2, 3])
    }
    model = YOLO("yolov8n.pt")
    results = model.train(data= f'{data_location}/data_change.yaml', project = 'hyperparameters', epochs=8, **params)
    print(params)
    print(f"mAP (0.5-0.95): {results.box.map}")
    print(f"mAP@0.5: {results.box.map50}")
````

````python
# pip install onnxruntime

from ultralytics import YOLO

# Load the YOLO model
model = YOLO("./runs/detect/yolov8/experiment/weights/best.pt") 

# Export the model to ONNX format
model.export(format="onnx")
````

````python
import onnxruntime as rt
import numpy as np
import glob

# Load the ONNX model
session = rt.InferenceSession("./runs/detect/yolov8/experiment/weights/best.onnx")

# Get input and output names
input_name = session.get_inputs()[0].name
output_names = [output.name for output in session.get_outputs()]

# Define the target input size for the ONNX model
input_size = (640, 640)

# Define the base path where the folders are located
base_path = '../notebooks-data/tyre-2/test/images'

image_path = glob.glob(f'{base_path}/*.jpg')[5]
image = cv2.imread(image_path)

# Resize the image while maintaining aspect ratio and padding
# This is a common method to prepare images for YOLO models
h, w = image.shape[:2]
scale = min(input_size[0] / h, input_size[1] / w)
resized_h, resized_w = int(h * scale), int(w * scale)
resized_image = cv2.resize(image, (resized_w, resized_h), interpolation=cv2.INTER_AREA)

padded_image = np.full((input_size[0], input_size[1], 3), 114, dtype=np.uint8) # Pad with a neutral color (e.g., gray)
x_offset = (input_size[1] - resized_w) // 2
y_offset = (input_size[0] - resized_h) // 2
padded_image[y_offset:y_offset + resized_h, x_offset:x_offset + resized_w] = resized_image

# Convert BGR to RGB (if your model expects RGB)
rgb_image = cv2.cvtColor(padded_image, cv2.COLOR_BGR2RGB)

# Normalize pixel values to 0-1 range
normalized_image = rgb_image.astype(np.float32) / 255.0

# Transpose to CHW (Channels First) and add batch dimension
# Assuming the ONNX model expects (1, 3, 640, 640)
input_tensor = np.transpose(normalized_image, (2, 0, 1)) # HWC to CHW
input_tensor = np.expand_dims(input_tensor, axis=0) # Add batch dimension

outputs = np.array(session.run(None, {input_name: input_tensor})[0])[0]
print(outputs.shape)
print(outputs[:,500])
 
````
