اهم دستورات مورد استفاده به منظور ذخیره‌سازی و بارگذاری مجدد مدل:

- save/load
- model.save_state_dict/model.load_state_dict
</div>

---

```python
import torch
from torch import nn

class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Linear(512, 10),
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits

model = Model()
```

---

<div dir="rtl" style="text-align: right;">
رویکرد اول: ذخیره وزن‌ها مستقل از ساختار
</div>

```python
torch.save(model.state_dict(), '../checkpoints/model_weights.pth')
```

```python
model = Model()
model.load_state_dict(torch.load('../checkpoints/model_weights.pth'))   # load: map_location=torch.device('cpu')
print(model)
```

```python
checkpoint = {
    'state_dict': model.state_dict(),
    'best_loss': 0.0001
}
torch.save(checkpoint, '../checkpoints/model_checkpoint.pth')
```

```python
checkpoint = torch.load('../checkpoints/model_checkpoint.pth')

model = Model()

model.load_state_dict(checkpoint['state_dict'])
```
