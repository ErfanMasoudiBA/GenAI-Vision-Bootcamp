شِمای کلی مدل، انتقال آن به دستگاه مورد نظر و یک بار عبور داده از مدل:
</div>

````python
import torch
from torch import nn


class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Linear(512, 10),
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits

model = Model()
````

````python
print(model)
````

````python
x = torch.randn(1, 28, 28)
y = model(x)
print(y.shape)
````

````python
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using {device} device")
# number
````

````python
model = Model().to(device)
x = torch.randn(1, 28, 28).to(device)
y = model(x)
print(y.shape)
````

<div dir="rtl" style="text-align: right;">
بررسی وضعیت پارامترهای مدل:
</div>

````python
from torch import nn

class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 10),
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits

model = Model()
for name, param in model.named_parameters():    # parameters and requires_grad
    print(f"Layer: {name} | Size: {param.size()}\n")
````
