<div dir="rtl" style="text-align: right;">
محاسبه خودکار مشتق به ازای هر گراف محاسباتی مبتنی بر تابع از پیش‌ساخته‌شده در پایتورچ به نام
torch.autograd
است.
</div>

---

```python
import torch

x = torch.ones(5)
y = torch.zeros(3)
w = torch.randn(5, 3, requires_grad=True)   # w.requires_grad
b = torch.randn(3, requires_grad=True)
z = torch.matmul(x, w) + b
loss = torch.nn.functional.cross_entropy(z, y)
```

<div dir="rtl" style="text-align: right;">
نکته۱: وجود گرادیان تنها به ازای گره‌هایی که پارامترگرادیان‌شان مثبت شده باشد.
</div>

```python
loss.backward()
print(w.grad)
print(b.grad)
```

<div dir="rtl" style="text-align: right;">
نکته۲: محاسبات backward به طور پیش‌فرض یک مرحله‌ای است، مگر با فرض استفاده از retain_graph.
</div>

```python
import torch

x = torch.ones(5)
y = torch.zeros(3)
w = torch.randn(5, 3, requires_grad=True)   # w.requires_grad
b = torch.randn(3, requires_grad=True)
z = torch.matmul(x, w) + b
loss = torch.nn.functional.cross_entropy(z, y)
loss.backward()
```

```python
loss.backward()
```

<div dir="rtl" style="text-align: right;">
پارامتر
retain_graph=True
را در تابع
backward
اول قرار دهید و سپس وضعیت فراخوانی
backward
دوم را رصد نمایید.
</div>

<div dir="rtl" style="text-align: right;">
عدم نیاز به ذخیره‌سازی گراف محاسباتی حین استنتاج:

- with torch.no_grad()
- detach
</div>

```python
import torch

x = torch.ones(2, 2, requires_grad=True)
w = torch.eye(2, requires_grad=True)
b = torch.zeros(2)

z = torch.matmul(x, w) + b
print(z.requires_grad)

with torch.no_grad():
    z = torch.matmul(x, w) + b
print(z.requires_grad)
```

```python
import torch

x = torch.ones(2, 2, requires_grad=True)
w = torch.eye(2, requires_grad=True)
b = torch.zeros(2)

z = torch.matmul(x, w) + b
print(z.requires_grad)

z_det = z.detach()
print(z_det.requires_grad)
```

<div dir="rtl" style="text-align: right;">
سوال:
کد زیر را بررسی نمایید. آیا backward برای y_detached قابل انجام است؟
</div>

```python
import torch

x = torch.randn(3, 3, requires_grad=True)
y = x * 2
y_detached = y.detach()
y_detached += 3
y.sum().backward()
```
