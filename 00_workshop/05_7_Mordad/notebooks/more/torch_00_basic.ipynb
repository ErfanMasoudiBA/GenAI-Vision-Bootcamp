```python
# we create tensors in a similar way to numpy nd arrays
x_numpy = np.array([0.1, 0.2, 0.3])
x_torch = torch.tensor([0.1, 0.2, 0.3])
print('x_numpy:', x_numpy )
print('x_torch:', x_torch )
```

```python
# to and from numpy, pytorch
# The tensor and ndarray share the same memory. 
print('to and from numpy and pytorch')
print('from numpy:', torch.from_numpy(x_numpy))
print('to numpy:', x_torch.numpy() )
```

```python
# we can do basic operations like +-*/
y_numpy = np.array([3,4,5.])
y_torch = torch.tensor([3,4,5.])
print("x+y")
print(x_numpy + y_numpy, x_torch + y_torch)
# print(x_numpy + y_torch) # TypeError
```

```python
# many functions that are in numpy are also in pytorch
print("numpy norm:", np.linalg.norm(x_numpy))
print("torch norm:", torch.norm(x_torch))
```

```python
# to apply an operation along a dimension,
# we use the dim keyword argument instead of axis
print("mean along the 0th dimension")
x_numpy = np.array([[1,2],[3,4.]])
x_torch = torch.tensor([[1,2],[3,4.]])
print("numpy mean:", np.mean(x_numpy, axis=0))
print("torch mean:", torch.mean(x_torch, dim=0))
```

```python
a = torch.ones(2, 2)
b = a.clone()

assert b is not a      # different objects in memory...
print(torch.eq(a, b))  # ...but still with the same contents!

a[0][1] = 561          # a changes...
print(b)  
```

```python
import torch
from torch import linalg
```

```python
# define a matrix with Tensor
A=torch.Tensor([[1,2],[3,4]])
B=torch.Tensor([[3,1],[2,5]])
print('A:',A.tolist())
print('B:',B.tolist())
```

```python
print('Determinant of A:',linalg.det(A).item())
print('Inverse of A:\n',linalg.inv(A))
print('A* A^-1:\n',(A@linalg.inv(A)))
```

```python
# a square matrix and its eigenvalues and eigenvectors
A=torch.Tensor([[1,2],[0,3]])
print('The square matrix:',A.tolist())
```

```python
# getting eigenvalues and eigenvectors
eigvals,eigvecs=linalg.eig(A)
print('Eigenvalues:',eigvals.tolist())
print('Eigenvectors:\n',eigvecs.tolist())
```

```python
print('Sum of eigenvalues:',eigvals.sum().item())
print('Sum of digonal elements:',A.diag().sum().item())
```

```python
print('Product of eigenvalues:',eigvals.prod().item())
print('Determinant of matrix:',A.det().item())
```
