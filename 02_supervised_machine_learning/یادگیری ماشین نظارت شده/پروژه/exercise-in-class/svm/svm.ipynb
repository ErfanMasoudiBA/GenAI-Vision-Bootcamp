```python
import pandas as pd
import numpy as np

from sklearn.svm import SVC
from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
```

```python
raw_data = pd.read_csv("./blood+transfusion+service+center/transfusion.data").set_axis(labels=list(range(5)),axis=1)
raw_data
```

```python
X = raw_data
y = np.array(X.pop(4))
X = np.array(X)
X
```

```python
min_max_scaler = MinMaxScaler()
X = min_max_scaler.fit_transform(X)
X
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,y, test_size=0.2)
print(X_train.shape, X_test.shape, Y_train.shape, Y_test.shape)
```

```python
svc_linear = SVC(kernel="linear")
svc_linear.fit(X_train,Y_train)
print(svc_linear.score(X_test, Y_test))
```

```python
svc_poly = SVC(kernel="poly", degree=5)
svc_poly.fit(X_train,Y_train)
print(svc_poly.score(X_test, Y_test))
```

```python
svc_rbf = SVC(kernel="rbf")
svc_rbf.fit(X_train,Y_train)
print(svc_rbf.score(X_test, Y_test))
```
