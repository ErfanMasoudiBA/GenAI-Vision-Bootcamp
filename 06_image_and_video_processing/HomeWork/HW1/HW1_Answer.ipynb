```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

# section 01

```python
BrightImage = "./Input_Images/BrightImage.jpg"
image = Image.open(BrightImage).convert("L")
BrightImage_8bit = np.array(image, dtype=np.uint8)

print(BrightImage_8bit.dtype)
print(BrightImage_8bit.min(),BrightImage_8bit.max())

plt.figure(figsize=(8, 6))
plt.imshow(BrightImage_8bit, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

```python
BrightImage_8bit
```

```python
I_BI = BrightImage_8bit
img = Image.fromarray(I_BI - 40)
img.save("./Output_Images/Part_1/I_BI-40.jpg")
img = Image.fromarray(I_BI - 20)
img.save("./Output_Images/Part_1/I_BI-20.jpg")
img = Image.fromarray(I_BI + 40)
img.save("./Output_Images/Part_1/I_BI+40.jpg")
img = Image.fromarray(I_BI + 20)
img.save("./Output_Images/Part_1/I_BI+20.jpg")
```

```python
vals = I_BI.flatten()
plt.hist(vals, 256, [0, 256])
plt.title("Histogram of BrightImage.jpg")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim([0, 256])
plt.show()
```

```python
mean_val_BI = int(I_BI.flatten().mean())
mean_val_BI
```

```python
b_BI = 128 - mean_val_BI
b_BI = b_BI*2
b_BI
```

```python
temp_img = I_BI.astype(np.int16) + b_BI
temp_img_clipped = np.clip(temp_img, 0, 255)
new_image_BI = temp_img_clipped.astype(np.uint8)


img = Image.fromarray(new_image_BI)
img.save("./Output_Images/Part_1/I_BI+b_BI.jpg")

plt.figure(figsize=(8, 6))
plt.imshow(new_image_BI, cmap="gray", vmin=0, vmax=255)
plt.axis("off")
plt.show()
```

```python
vals = (new_image_BI).flatten()
plt.hist(vals, 256, [0, 256])
plt.title("Histogram of DarkImage.jpg")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim([0, 256])
plt.show()
```

# section 02

```python
multipliers = [1, 0.5, 0.75, 1.25, 1.5]

for m in multipliers:
    img_data = np.clip(I_DI * m, 0, 255).astype(np.uint8)
    img = Image.fromarray(img_data)
    img.save(f"./Output_Images/Part_1/I_DIx{m}.jpg")
    img_data = np.clip(I_BI * m, 0, 255).astype(np.uint8)
    img = Image.fromarray(img_data)
    img.save(f"./Output_Images/Part_1/I_BIx{m}.jpg")
```

```python
fig, axes = plt.subplots(2, len(multipliers), figsize=(15, 6))

for i, m in enumerate(multipliers):
    processed_BI = np.clip(I_BI * m, 0, 255).astype(np.uint8)
    axes[0, i].imshow(processed_BI, cmap='gray', vmin=0, vmax=255) 
    axes[0, i].set_title(f'Bright x {m}')
    axes[0, i].axis('off')

    processed_DI = np.clip(I_DI * m, 0, 255).astype(np.uint8)
    axes[1, i].imshow(processed_DI, cmap='gray', vmin=0, vmax=255)
    axes[1, i].set_title(f'Dark x {m}')
    axes[1, i].axis('off')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, len(multipliers), figsize=(20, 10))

for i, m in enumerate(multipliers):
    
    di_vals = np.clip(I_DI * m, 0, 255).astype(np.uint8).flatten()
    axes[0, i].hist(di_vals, bins=256, range=[0, 256], color='gray')
    axes[0, i].set_title(f"Dark Image x {m}")
    axes[0, i].set_xlim([0, 256])
    
    bi_vals = np.clip(I_BI * m, 0, 255).astype(np.uint8).flatten()
    axes[1, i].hist(bi_vals, bins=256, range=[0, 256], color='gray')
    axes[1, i].set_title(f"Bright Image x {m}")
    axes[1, i].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
# 64 = 256/4
a_BI = np.ceil(64/I_BI.std())
a_DI = np.ceil(64/I_DI.std())
a_BI ,a_DI
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

processed_DI = np.clip(I_DI * a_DI, 0, 255).astype(np.uint8)
axes[1, 0].hist(di_vals, bins=256, range=[0, 256], color='gray')
axes[1, 0].set_title("Dark Image")
axes[1, 0].set_xlim([0, 256])

processed_DI = np.clip(I_DI* a_DI, 0, 255).astype(np.uint8)
axes[1, 1].hist(di_vals, bins=256, range=[0, 256], color='gray')
axes[1, 1].set_title(f"Dark Image x {a_DI}")
axes[1, 1].set_xlim([0, 256])

processed_BI = np.clip(I_BI, 0, 255).astype(np.uint8)
axes[0, 0].hist(bi_vals, bins=256, range=[0, 256], color='gray')
axes[0, 0].set_title("Bright Image")
axes[0, 0].set_xlim([0, 256])

processed_BI = np.clip(I_BI / a_BI, 0, 255).astype(np.uint8)
axes[0, 1].hist(bi_vals, bins=256, range=[0, 256], color='gray')
axes[0, 1].set_title(f"Bright Image / {a_BI}")
axes[0, 1].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 6))

processed_DI = np.clip(I_DI, 0, 255).astype(np.uint8)
axes[1, 0].imshow(processed_DI, cmap='gray', vmin=0, vmax=255)
axes[1, 0].set_title("Dark Image")
axes[1, 0].axis('off') 

processed_DI_scaled = np.clip(I_DI * a_DI, 0, 255).astype(np.uint8)
axes[1, 1].imshow(processed_DI_scaled, cmap='gray', vmin=0, vmax=255)
axes[1, 1].set_title(f"Dark Image x {a_DI}")
axes[1, 1].axis('off')

processed_BI = np.clip(I_BI, 0, 255).astype(np.uint8)
axes[0, 0].imshow(processed_BI, cmap='gray', vmin=0, vmax=255)
axes[0, 0].set_title("Bright Image")
axes[0, 0].axis('off')

processed_BI_scaled = np.clip(I_BI / a_BI, 0, 255).astype(np.uint8)
axes[0, 1].imshow(processed_BI_scaled, cmap='gray', vmin=0, vmax=255)
axes[0, 1].set_title(f"Bright Image / {a_BI}")
axes[0, 1].axis('off')

plt.tight_layout()
plt.show()
```

# section 03

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 6))

I_float = I_DI.astype(np.float32)

mu_old = I_float.mean()
std_old = I_float.std()

a_opt = 64 / std_old
b_opt = 128 - (a_opt * mu_old)

I_new = (a_opt * I_float) + b_opt

new_DI = np.clip(I_new, 0, 255).astype(np.uint8)

axes[0].imshow(new_DI, cmap='gray', vmin=0, vmax=255)
axes[0].set_title("New Dark Image")
axes[0].axis('off')

I_float = I_BI.astype(np.float32)

mu_old = I_float.mean()
std_old = I_float.std()

a_opt = 64 / std_old
b_opt = 128 - (a_opt * mu_old)

I_new = (a_opt * I_float) + b_opt

new_BI = np.clip(I_new, 0, 255).astype(np.uint8)

axes[1].imshow(new_BI, cmap='gray', vmin=0, vmax=255)
axes[1].set_title(f"New Bright Image")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1,2, figsize=(15, 5))

processed_DI = np.clip(new_DI, 0, 255).astype(np.uint8)
axes[1].hist(di_vals, bins=256, range=[0, 256], color='gray')
axes[1].set_title("New Dark Image Hist")
axes[1].set_xlim([0, 256])

processed_BI = np.clip(new_BI, 0, 255).astype(np.uint8)
axes[0].hist(bi_vals, bins=256, range=[0, 256], color='gray')
axes[0].set_title("New Bright Image Hist")
axes[0].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, len(gammas), figsize=(20, 10))

for i, m in enumerate(gammas):
    
    di_vals = np.clip(I_DI ** m, 0, 255).astype(np.uint8).flatten()
    axes[0, i].hist(di_vals, bins=256, range=[0, 256], color='gray')
    axes[0, i].set_title(f"Dark Image ^ {m}")
    axes[0, i].set_xlim([0, 256])
    
    bi_vals = np.clip(I_BI ** m, 0, 255).astype(np.uint8).flatten()
    axes[1, i].hist(bi_vals, bins=256, range=[0, 256], color='gray')
    axes[1, i].set_title(f"Bright Image ^ {m}")
    axes[1, i].set_xlim([0, 256])

plt.tight_layout()
plt.show()
```

```python
gammas = [1, 0.8, 0.9, 1.1, 1.2]
fig, axes = plt.subplots(2, len(gammas), figsize=(15, 6))

for i, m in enumerate(gammas):
    I_BI_norm = I_BI.astype(np.float32) / 255.0
    I_BI_gamma = I_BI_norm ** m
    processed_BI = np.clip(I_BI_gamma * 255.0, 0, 255).astype(np.uint8)
    axes[0, i].imshow(processed_BI, cmap='gray', vmin=0, vmax=255) 
    axes[0, i].set_title(f'Bright ^ {m}')
    axes[0, i].axis('off')
    
    I_DI_norm = I_DI.astype(np.float32) / 255.0
    I_DI_gamma = I_DI_norm ** m
    processed_DI = np.clip(I_DI_gamma * 255.0, 0, 255).astype(np.uint8)
    axes[1, i].imshow(processed_DI, cmap='gray', vmin=0, vmax=255)
    axes[1, i].set_title(f'Dark ^ {m}')
    axes[1, i].axis('off')
    
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, len(gammas), figsize=(20, 6))

for i, m in enumerate(gammas):
    
    I_BI_norm = I_BI.astype(np.float32) / 255.0
    I_BI_gamma = I_BI_norm ** m
    processed_BI = np.clip(I_BI_gamma * 255.0, 0, 255).astype(np.uint8)
    
    axes[0, i].hist(processed_BI.flatten(), bins=256, range=[0, 256], color='gray') 
    axes[0, i].set_title(f'Bright ^ {m}')
    axes[0, i].set_xlim([0, 256])
    
    I_DI_norm = I_DI.astype(np.float32) / 255.0
    I_DI_gamma = I_DI_norm ** m
    processed_DI = np.clip(I_DI_gamma * 255.0, 0, 255).astype(np.uint8)
    
    axes[1, i].hist(processed_DI.flatten(), bins=256, range=[0, 256], color='gray')
    axes[1, i].set_title(f'Dark ^ {m}')
    axes[1, i].set_xlim([0, 256])
    
plt.tight_layout()
plt.show()
```

# section 05

```python
import cv2

I_DI_eq = cv2.equalizeHist(I_DI)
I_BI_eq = cv2.equalizeHist(I_BI)
```

```python
fig_img, axes_img = plt.subplots(2, 3, figsize=(15, 10))
fig_img.suptitle("Images Comparison: Original vs Equalized vs Linear Stretched", fontsize=16, fontweight='bold')

axes_img[0, 0].imshow(I_DI, cmap='gray', vmin=0, vmax=255)
axes_img[0, 0].set_title('Original Dark Image')

axes_img[0, 1].imshow(I_DI_eq, cmap='gray', vmin=0, vmax=255)
axes_img[0, 1].set_title('Equalized Dark Image')

axes_img[0, 2].imshow(new_DI, cmap='gray', vmin=0, vmax=255)
axes_img[0, 2].set_title('New Dark Image (Stretched)')

axes_img[1, 0].imshow(I_BI, cmap='gray', vmin=0, vmax=255)
axes_img[1, 0].set_title('Original Bright Image')

axes_img[1, 1].imshow(I_BI_eq, cmap='gray', vmin=0, vmax=255)
axes_img[1, 1].set_title('Equalized Bright Image')

axes_img[1, 2].imshow(new_BI, cmap='gray', vmin=0, vmax=255)
axes_img[1, 2].set_title('New Bright Image (Stretched)')

for ax in axes_img.flat:
    ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
fig_hist, axes_hist = plt.subplots(2, 3, figsize=(18, 10))
fig_hist.suptitle("Histograms Comparison: Original vs Equalized vs Linear Stretched", fontsize=16, fontweight='bold')

hist_kwargs = {'bins': 256, 'range': [0, 256], 'color': 'gray'}

axes_hist[0, 0].hist(I_DI.flatten(), **hist_kwargs)
axes_hist[0, 0].set_title('Original Hist (DI)')

axes_hist[0, 1].hist(I_DI_eq.flatten(), bins=256, range=[0, 256], color='black') # رنگ متفاوت برای تنوع
axes_hist[0, 1].set_title('Equalized Hist (DI)')

axes_hist[0, 2].hist(new_DI.flatten(), **hist_kwargs)
axes_hist[0, 2].set_title('New Hist (DI Stretched)')

axes_hist[1, 0].hist(I_BI.flatten(), **hist_kwargs)
axes_hist[1, 0].set_title('Original Hist (BI)')

axes_hist[1, 1].hist(I_BI_eq.flatten(), bins=256, range=[0, 256], color='black')
axes_hist[1, 1].set_title('Equalized Hist (BI)')

axes_hist[1, 2].hist(new_BI.flatten(), **hist_kwargs)
axes_hist[1, 2].set_title('New Hist (BI Stretched)')

for ax in axes_hist.flat:
    ax.set_xlim([0, 256])

plt.tight_layout()
plt.show()
```
